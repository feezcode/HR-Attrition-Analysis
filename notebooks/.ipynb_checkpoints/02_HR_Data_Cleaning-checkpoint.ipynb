```python
import pandas as pd

df = pd.read_csv('/Users/alhafeez/Documents/Porto-Tableau/HR-Analysis/data/HRDataset_v14.csv')
```

```python
text_cols = df.select_dtypes(include="object").columns
df[text_cols] = df[text_cols].apply(lambda x: x.str.strip())
```

```python
df["HispanicLatino"] = df["HispanicLatino"].str.title()
```

```python
df[date_cols] = df[date_cols].apply(pd.to_datetime, format="mixed")
```

```python
df["ManagerID"] = df["ManagerID"].fillna(-1).astype(int)
```

```python
df.info()
df.isnull().sum()
```

```python
df.to_csv("HRDataset_cleaned.csv", index=False)
```
